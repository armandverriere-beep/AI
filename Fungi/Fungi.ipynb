 # How many of P class fungi have no odor?

 ### I will conduct my research by analyzing the data in the table and add up the mushrooms that are both class P and have no odor.

```python
import pandas as pd
```

```python
fungi = pd.read_csv("mushrooms.csv")
```

```python
fungi.info()
```

```python
fungi["odor"]
```

```python
fungi["odor"].value_counts()
```

```python
clas = (fungi[fungi["class"] == "p"])
```

```python
clas['odor'].value_counts()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=clas, x='odor', order=clas['odor'].value_counts().index)
plt.title('Odor Distribution in Class P Mushrooms')
plt.xlabel('Odor type')
plt.ylabel('Count')
plt.show()
```

The frequency of n is approximately 120. It means that in the poisonous category, out of 3340 poisonous mushrooms, only 120 have no odors.
